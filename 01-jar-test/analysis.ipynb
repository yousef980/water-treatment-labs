# 01 · Coagulation-Flocculation Optimization via Jar-Testing

## Objective
Determine the optimum coagulant (Aluminum Sulfate) dose for maximum turbidity removal efficiency.

## Key Parameters
- **Matrix Volume:** 500 mL raw surface water per beaker
- **Coagulant:** 10 g/L Aluminum Sulfate (Al₂(SO₄)₃)
- **Rapid Mixing:** 100 rpm for 2 minutes
- **Slow Mixing:** 30 rpm for 20 minutes
- **Sedimentation:** Static rest for 20 minutes

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '..')
from utils.analysis_tools import turbidity_removal_efficiency

# Experimental data
data = {
    'Beaker': ['B1', 'B2', 'B3 (Optimum)', 'B4', 'B5', 'B6'],
    'Dose_mg_L': [10, 20, 30, 40, 50, 60],
    'Residual_Turbidity_NTU': [14.2, 8.5, 1.8, 4.3, 9.1, 15.4]
}

df = pd.DataFrame(data)
print(df)
print(f"\nDataFrame shape: {df.shape}")
```

```python
# Calculate removal efficiency
# Initial turbidity (worst case, beaker B1)
initial_turbidity = df['Residual_Turbidity_NTU'].iloc[0]

df['Removal_Efficiency_%'] = df['Residual_Turbidity_NTU'].apply(
    lambda x: turbidity_removal_efficiency(initial_turbidity, x)
)

print(df[['Dose_mg_L', 'Residual_Turbidity_NTU', 'Removal_Efficiency_%']])

# Find optimal dose
optimal_idx = df['Residual_Turbidity_NTU'].idxmin()
optimal_dose = df.loc[optimal_idx, 'Dose_mg_L']
optimal_turbidity = df.loc[optimal_idx, 'Residual_Turbidity_NTU']
optimal_efficiency = df.loc[optimal_idx, 'Removal_Efficiency_%']

print(f"\n" + "="*60)
print(f"OPTIMAL COAGULANT DOSE: {optimal_dose} mg/L")
print(f"Residual Turbidity: {optimal_turbidity} NTU")
print(f"Removal Efficiency: {optimal_efficiency:.2f}%")
print("="*60)
```

```python
# Plot: Turbidity Removal Curve
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Turbidity curve
ax1.plot(df['Dose_mg_L'], df['Residual_Turbidity_NTU'], 
         color='#38BDF8', linewidth=2.5, marker='o', markersize=8, label='Residual Turbidity')
ax1.scatter([optimal_dose], [optimal_turbidity], color='#10B981', s=200, marker='v', 
           zorder=5, label=f'Optimum ({optimal_dose} mg/L)')
ax1.set_xlabel('Al₂(SO₄)₃ Dose (mg/L)', fontweight='bold')
ax1.set_ylabel('Residual Turbidity (NTU)', fontweight='bold')
ax1.set_title('Turbidity Removal Curve', fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Removal efficiency curve
ax2.plot(df['Dose_mg_L'], df['Removal_Efficiency_%'], 
         color='#8B5CF6', linewidth=2.5, marker='s', markersize=8)
ax2.scatter([optimal_dose], [optimal_efficiency], color='#10B981', s=200, marker='v', 
           zorder=5, label='Peak Efficiency')
ax2.set_xlabel('Al₂(SO₄)₃ Dose (mg/L)', fontweight='bold')
ax2.set_ylabel('Removal Efficiency (%)', fontweight='bold')
ax2.set_title('Removal Efficiency vs Dose', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.savefig('jar_test_analysis.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"Chart saved as 'jar_test_analysis.png'")
```

## Technical Interpretation

### Zone 1: Under-Dosing (10–20 mg/L)
- **Observation:** Residual turbidity remains high (14.2 → 8.5 NTU)
- **Mechanism:** Insufficient Al³⁺ cations to compress the electric double layer around colloids
- **Result:** Electrostatic repulsion prevents aggregation

### Zone 2: Optimal Window (30 mg/L)
- **Observation:** Minimum turbidity achieved (1.8 NTU)
- **Mechanism:** Perfect charge neutralization → maximum floc formation
- **Economic significance:** Lowest chemical cost + best clarification

### Zone 3: Over-Dosing (40–60 mg/L)
- **Observation:** Turbidity **increases** back to 15.4 NTU
- **Mechanism:** Colloidal **restabilization** — excess Al³⁺ reverses surface charges
- **Industrial implication:** Waste of reagent + poor treatment outcome
